# Neue CSV-Datei erstellen
Es looped durch jeden Eitrag im vorher definierten Array von der Start- zu Endnummer und erstellt eine Zeile mit einer spezifischen PLZ die dem jeweiligen Bundesland angehört.

Einige PLZ Einträge sind in dieser Liste Bundesländern zugeordnet, obwohl diese in Realität keinem Bundesland zugeordnet sind. Das hat aber keinen negativen Einfluss auf die Ergebnisse der Zuordnung, da nur echte PLZ Nummern aus der API abgerufen werden.

```python
with open('Liste-der-PLZ.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile, delimiter=',')
    writer.writerow(['PLZ', 'Bundesland'])

    for plz_von, plz_bis, bundesland in plz_bereiche:
        start = int(plz_von)
        ende = int(plz_bis)

        for plz_num in range(start, ende + 1):
            plz = f"{plz_num:05d}"
            writer.writerow([plz, bundesland])

print(f"PLZ wurden in die CSV geschrieben")
```
