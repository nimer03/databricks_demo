```sql
SELECT
	'Alle Kunden' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
UNION
SELECT
	'Ohne Null' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer
WHERE
	plz IS NOT NULL
UNION
SELECT
	'Ohne Fehler' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address LIKE concat('%', c.plz, '%');

SELECT
	c.plz,
	b.address
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address NOT LIKE concat('%', c.plz, '%')
	AND c.plz IS NOT NULL
ORDER BY
	c.customer_id;
```
