# Analyse von Connection Log
Überprüfen auf Outlier und fehlerhafte String-Werte

```python
df_log = spark.table(catalog + '.' + schema + '.' + 'log_bronze')

num_columns = ['speed_measured_mbps', 'packet_loss_percent', 'latency_ms', 'downtime_minutes', 'connection_drops_count']

for column in num_columns:
    df_log = df_log.withColumn(column, col(column).cast('double'))

display(df_log)
```

# Analyse von Chat Transkript
Überprüfen, ob das JSON-Format eingehalten wird und ob fehlerhafte Werte vorliegen

```python


df_chat = spark.table(catalog + '.' + schema + '.' + 'chat_bronze')

json_schema = ArrayType(StructType([StructField("name", StringType(), nullable = True), StructField("value", StringType(), nullable = True)]))

df = df_chat\
  .withColumn("correct_json", from_json(col("messages"), json_schema))\
  .withColumn("json_flag", when(col("messages").isNotNull() & col("correct_json").isNull(), 1).otherwise(0))\
  .drop("correct_json")

display(df)
```
