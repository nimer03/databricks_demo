```sql
-- Create chat table
CREATE TABLE IF NOT EXISTS sac.support.chat (
		session_id STRING NOT NULL,
		customer_id STRING,
		agent_id STRING,
		classification STRING,
		comment STRING,
		CONSTRAINT chat_pk PRIMARY KEY (session_id),
		CONSTRAINT chat_f1 FOREIGN KEY (customer_id) REFERENCES sac.customer.customer (customer_id),
		CONSTRAINT chat_f2 FOREIGN KEY (agent_id) REFERENCES sac.employee.agent (agent_id)
	);

-- Edit constraints of table
ALTER TABLE
	sac.support.chat
DROP CONSTRAINT IF EXISTS
	chat_transcript_ci;

ALTER TABLE
	sac.support.chat
DROP CONSTRAINT IF EXISTS
	chat_transcript_ai;

ALTER TABLE
	sac.support.chat
ADD
	CONSTRAINT chat_transcript_ci CHECK (customer_id IS NOT NULL);

ALTER TABLE
	sac.support.chat
ADD
	CONSTRAINT chat_transcript_ai CHECK (agent_id IS NOT NULL);

-- Create message table
CREATE TABLE IF NOT EXISTS sac.support.message (
		session_id STRING NOT NULL,
		speaker STRING,
		timestamp TIMESTAMP NOT NULL,
		message STRING,
		sentiment STRING,
		CONSTRAINT message_fk FOREIGN KEY (session_id) REFERENCES sac.support.chat (session_id),
		CONSTRAINT message_pk PRIMARY KEY (session_id, speaker, timestamp)
	);

-- Edit constraints of table
ALTER TABLE
	sac.support.message
DROP CONSTRAINT IF EXISTS
	message_ts;

ALTER TABLE
	sac.support.message
ADD
	CONSTRAINT message_ts
		CHECK (
			timestamp IS NOT NULL
			AND timestamp <= current_timestamp()
		);
```
