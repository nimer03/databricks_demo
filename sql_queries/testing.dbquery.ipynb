```sql
-- DB_11: Valide Änderung
SELECT
	*
FROM
	sac.support.ticket
WHERE
	ticket_id = "TKT_000003";

UPDATE
	sac.support.ticket_bronze
SET
	priority = "high"
WHERE
	ticket_id = "TKT_000003";

-- DB_12: Nicht valide Änderung
UPDATE
	sac.support.ticket_bronze
SET
	timestamp_created = "2026-06-30T04:59:57.000+00:00"
WHERE
	ticket_id = "TKT_000003";

-- DB_13: Löschen
DESCRIBE HISTORY sac.employee.agent;

-- DB_21: Constraint-Grenze
UPDATE
	sac.employee.agent
SET
	monthly_salary_eur = 0
WHERE
	agent_id = "AGENT_001";

-- DB_22: Fremschlüssel
DELETE FROM
	sac.customer.customer
WHERE
	(
		SELECT
			customer_id
		FROM
			sac.support.ticket
		WHERE
			ticket_id = "TKT_000003"
	) = customer_id;

-- DB_31: Analyse
SELECT
	m.timestamp,
	c.classification,
	m.sentiment
FROM
	sac.support.message m
		JOIN sac.support.chat c
			ON m.session_id = c.session_id
WHERE
	m.speaker = "customer"
	AND m.session_id = "CHAT_00479"
ORDER BY
	m.timestamp
```
