```sql
-- Summe aller Kunden, 
-- Summe der Kunden mit PLZ Eintrag, 
-- Summe der Kunden, die eine korrekte PLZ haben
SELECT
	'Alle Kunden' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
UNION
SELECT
	'Ohne Null' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer
WHERE
	plz IS NOT NULL
UNION
SELECT
	'Ohne Fehler' AS beschreibung,
	count(*) AS sum_cust
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address LIKE concat('%', RIGHT(c.plz, 4), '%');

-- PLZ ist nicht in der Originaladresse (Fehler ausgeschlossen)
SELECT
	plz,
	b.address
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address NOT LIKE concat('%', RIGHT(c.plz, 4), '%')
	AND b.address NOT LIKE '%, None%'
	AND plz IS NOT NULL
ORDER BY
	c.customer_id;

-- Stadt angegeben aber kein Bundesland
SELECT
	plz,
	city,
	zone,
	address
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	zone IS null
	and city != 'None';

-- PLZ mit alphabetischen Werten
SELECT
	plz,
	address
FROM
	customer_service.customer c
		JOIN customer_service.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	LEN(plz) > 5
```
