```sql
-- Summe aller Kunden,
-- Summe der Kunden mit PLZ Eintrag,
-- Summe der Kunden, die eine korrekte PLZ haben
SELECT
	'Alle Kunden' AS beschreibung,
	count(*) AS sum_cust
FROM
	sac.customer.customer c
		JOIN sac.customer.customer_bronze b
			ON c.customer_id = b.customer_id
UNION
SELECT
	'Ohne Null' AS beschreibung,
	count(*) AS sum_cust
FROM
	sac.customer.customer
WHERE
	plz IS NOT NULL
UNION
SELECT
	'Ohne Fehler' AS beschreibung,
	count(*) AS sum_cust
FROM
	sac.customer.customer c
		JOIN sac.customer.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address LIKE concat('%', right(c.plz, 4), '%');

-- PLZ ist nicht in der Originaladresse (Fehler ausgeschlossen)
SELECT
	plz,
	b.address
FROM
	sac.customer.customer c
		JOIN sac.customer.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	b.address NOT LIKE concat('%', right(c.plz, 4), '%')
	AND b.address NOT LIKE '%, None%'
	AND plz IS NOT NULL
ORDER BY
	c.customer_id;

-- Stadt angegeben aber kein Bundesland
SELECT
	plz,
	city,
	zone,
	address
FROM
	sac.customer.customer c
		JOIN sac.customer.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	zone IS NULL
	AND city != 'None';

-- PLZ mit alphabetischen Werten
SELECT
	plz,
	address
FROM
	sac.customer.customer c
		JOIN sac.customer.customer_bronze b
			ON c.customer_id = b.customer_id
WHERE
	len(plz) > 5;

-- Original fehlende PLZ vs. befüllt
SELECT
	"Summe der KI-Werte" as Bundeland,
	count(*) as Summe
FROM
	sac.customer.customer c
WHERE
	ai_used = TRUE
UNION
SELECT
	zone as Bundeland,
	count(*) as Summe
FROM
	sac.customer.customer c
WHERE
	ai_used = TRUE
GROUP BY
	zone
```
